Carga en variables de entorno la API de Gemini.

````python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

chat = ChatGoogleGenerativeAI(model="gemini-2.0-flash", api_key=api_key)
````

Importa la clase de `ChatGoogleGenerativeAI` y selecciona un modelo de los disponibles:

* Gemini 1.5 Flash
* Gemini 2.0 Flash
* Gemini 2.5 Flash

También puedes probar a jugar con distintos límites de respuesta, tokens, etc...

````python
modelo_elegido = "gemini-2.0-flash"
````

````python
chat = ChatGoogleGenerativeAI(
    model=modelo_elegido,
    api_key=api_key,
    max_tokens=256,
    temperature=0.7
)
````

    Usa la función invoke, indicando qué es mensaje para el sistema y qué es interacción o consulta humana.

````python
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "Eres un experto en Python, vas a ayudar al usuario con sus preguntas con claridad y precisión, extiende lo que dices a dos párrafos."),
    ("user", "{question}")
])

parser = StrOutputParser()

chain = prompt_template | chat | parser

````

````python
def interactuar_con_chatbot():
    while True:
        print("\nHola, introduce tu pregunta sobre Python :D")
        pregunta_usuario = input("¿Qué necesitas?: ")

        if pregunta_usuario.lower() == "salir":
            print("Sistema: ¡Chaito!")
            break

        try:
            response_obj = chat.invoke({"messages": [{"role": "user", "content": pregunta_usuario}]})
            texto_respuesta = response_obj.get('text', '')
            metadatos = {k: v for k, v in response_obj.items() if k != 'text'}

            print("\nChatbot:\n", texto_respuesta)

        except Exception as e:
            print("Error al obtener la respuesta:", e)
````

````python

interactuar_con_chatbot()
````
